```python
ket_zero=np.array([[1],[0]])
ket_zero
```

```python
ket_one=np.array([[0],[1]])
ket_one
```

```python
N=int(N)
qs = np.empty((1, N), dtype=object)

for i in range(N):
    qs[0, i] = ket_zero.copy()

qs
```

```python
# gates
def operation_by_gates(info,qs):
    i=0
    while(i<info.shape[0]):
        #Hadamard gate:
        if(info[i,0]=='H'):
            H=np.array(
                [1,1],
                [1,-1]
            )/np.sqrt(2)
            qs[int(info[i,0])]=H@qs[int(info[i,0])]

        #CNOT Gate:
        if(info[i,0]=='CNOT'):
            control=int(info[i,1][0])
            target=int(info[i,1][1])
            if(qs[control]==ket_zero):
                continue
            else:
                if(qs[target]==ket_zero):
                    qs[target]=ket_one
                if(qs[target]==ket_one):
                    qs[target]=ket_zero

        #Ry gate:
        if(info[i,0]=='RY'):
            qubit_no=info[i,1][0]
            theta=info[i,1][1]
            RY=np.array(
                [[np.cos(theta/2),-np.sin(theta/2)],
                 [np.sin(theta/2),np.cos(theta/2)]]
                 )
            qs[qubit_no]=RY@qs[qubit_no]

        #Rx gate:
        if(info[i,0]=='RX'):
            qubit_no=info[i,1][0]
            theta=info[i,1][1]
            RY=np.array(
                        [[np.cos(theta/2),-np.sin(theta/2)*1j],
                        [-np.sin(theta/2)*1j,np.cos(theta/2)]]
                        )
            qs[qubit_no]=RY@qs[qubit_no]
            
        #Rz gate:
        if(info[i,0]=='RZ'):
            qubit_no=info[i,1][0]
            theta=info[i,1][1]
            RY=np.array(
                        [[np.cos(theta/2)-np.sin(theta/2),0],
                        [0,np.cos(theta/2)+np.sin(theta/2)]]
                        )
            qs[qubit_no]=RY@qs[qubit_no]
             
        #X gate:
        if(info[i,0]=='X'):
            X=np.array(
                        [[0,1],
                        [1,0]]
                        )
            qs[int(info[i,0])]=X@qs[int(info[i,0])]
        #Y gate:
        if(info[i,0]=='Y'):
            Y=np.array([
                [0,-1j],
                [1j,0]
            ])
        #Z gate:
        if(info[i,0]=='Z'):
            Z=np.array([
                [1,0],
                [0,-1]
            ])
        #TOFFOLI gate/CCNOT gate:
        if(info[i,0]=='TOFFOLI'):
            if(info[i,0]=='TOFOLLI'):
                control1=info[i,1][0]
                control2=info[i,1][1]
                target=info[i,1][2]
                if(qs(control1)==ket_one and qs(control2)==ket_one):
                    if(qs[target]==ket_one):
                        qs[target]=ket_zero
                    else:
                        qs[target]=ket_one
                        
        #SWAP gate:
        if(info[i,0]=='SWAP'):
            if(info[i,1][0]!=info[i,1][1]):
                qs[info[i,1][0]],qs[info[i,1][1]]=qs[info[i,1][1]],qs[info[i,1][0]]
```
