```python
#Function to calculate Gini Impurity
def Gini(labels):
    if(len(labels)==0):
        return 0
    species_list,occurances=np.unique(labels,return_counts=True)
    #This function  returns 2 np.arrays 

    proportions=occurances/len(labels)

    squared_sum=0
    for p in proportions:
        squared_sum+=(p**2)

    return 1-squared_sum

# return 1 - np.sum(proportions ** 2) ALITER
```

```python
def decision_Tree(df):
    #base cases in recursion:
    if(len(df)<=20):
        names,counts=np.unique(df["species"].to_numpy(),return_counts=True)
        return str(names[np.argmax(counts)])
    if(df["species"].nunique()==1):
        return str(df["species"].iloc[0])
    
    best_gini=float("inf")
    best_j=None
    best_thershold=float("inf")
    final_treshold=None
    for j in range(X.shape[1]):
        avg=np.array([])
        Gini_array=np.array([])
        sorted_df=df.sort_values(by=df.columns[j])
        sorted_current_col=sorted_df.iloc[:,j]
        #current_col is  a series so by="" is not needed

        #This loop traverses through a col and each iteration corresponds
        #to each avearge we encounter
        for i in range(len(sorted_current_col)-1):
            if sorted_current_col.iloc[i] == sorted_current_col.iloc[i + 1]:
                continue
            #First calculate the i th average and store in an array
            thresh=(sorted_current_col.iloc[i]+sorted_current_col.iloc[i+1])/2
            avg=np.append(avg,thresh)
            #calculate gini impurity of left group
            label1=sorted_df["species"].iloc[:i+1].to_numpy()
            if(label1.size>0):
                g1=Gini(label1)
            #calculate gini impurity of right group
            label2=sorted_df["species"].iloc[i+1:].to_numpy()
            if(label2.size>0):
                g2=Gini(label2)
            #calculate weighted GI for i th avg and store in array
            Gini_array=np.append(Gini_array,((i+1)*g1 + (len(sorted_current_col)-i-1)*g2)/len(sorted_current_col))
        if len(Gini_array) == 0:
            continue
        min_GI=min(Gini_array)
        if(min_GI<best_gini):
            best_gini=min_GI
            best_j=j
            final_treshold=avg[np.argmin(Gini_array)]

    if best_j is None:
        names, counts = np.unique(df["species"], return_counts=True)
        return names[np.argmax(counts)] 
    
    best_sorted_df = df.sort_values(by=df.columns[best_j])
    #best_sorted_df is still a df
    k=0
    n=len(best_sorted_df)
    while(k<n and best_sorted_df.iloc[k,best_j]<=final_treshold):
        k+=1
    left_label=best_sorted_df.iloc[:k]
    right_label=best_sorted_df.iloc[k:]

    left_child=decision_Tree(left_label)
    right_child=decision_Tree(right_label)

    return {
        "feature": df.columns[best_j],
        "threshold":final_treshold,
        "left": left_child,
        "right": right_child
    }
#Only training data goes in this function.
#This function takes in input in the form of training df 
#And in output it builds the entire decision tree and 
#returns the dict of the root node.


    #print("best_gini=",best_gini)
    #print("best_j=",best_j)
    #print("final_treshold=",final_treshold)
```

```python
def predict_one(f,t):
    '''
f represenst a single flower which has its features like
petal length, sepal length etc. and these are stored in a series 
called f and we are passing this as parameter as we require
features to predict something
'''
#When t is a leaf:(base case)
    if(type(t)!=dict):
        return t
    
#When t is not a leaf:
    feature=t["feature"]
    threshold = t["threshold"]

    if f[feature] <= threshold:
        return predict_one(f, t["left"])
    else:
        return predict_one(f, t["right"])
    
```
